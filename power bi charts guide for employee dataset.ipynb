# 📊 Power BI  Guide: Bar Chart (Employees by Gender)

## 🎯 Objective
Create a **Bar Chart** to show the **Number of Employees by Gender (Male vs Female)**.

---

## 🔹 Step 1: Open Power BI
1. Launch **Power BI Desktop**
2. Wait for the Home screen to load


---

## 🔹 Step 2: Load Dataset
1. Click on **Home tab**
2. Select **Get Data**
3. Choose **Text/CSV**
4. Browse and select your file: `employee_data.csv`
5. Click **Open**
6. Preview window will appear
7. Click **Load**

---

## 🔹 Step 3: Verify Data Loaded
1. Go to the **Data View** (table icon on left side)
2. Ensure columns are visible:
   - EmployeeID
   - EmployeeName
   - Gender
   - Age
   - Department
   - JobRole
   - Location

✔ If data is visible → proceed

---

## 🔹 Step 4: Go to Report View
1. Click on **Report View** (chart icon on left side)

---

## 🔹 Step 5: Insert Bar Chart Visual
1. Go to **Visualizations Pane** (right side)
2. Click on **Clustered Bar Chart** icon

✔ A blank chart will appear on canvas

---

## 🔹 Step 6: Add Fields to Chart

### ➤ Add Axis (Category)
1. Drag **Gender** column
2. Drop into **Y-axis**

### ➤ Add Values (Measure)
1. Drag **EmployeeID**
2. Drop into **X-axis**

---

## 🔹 Step 7: Set Aggregation (VERY IMPORTANT)
1. In the chart fields section
2. Click dropdown next to **EmployeeID**
3. Select **Count**

✔ This ensures number of employees is counted correctly

---

## 🔹 Step 8: Resize the Chart
1. Click on chart
2. Drag corners to adjust size
3. Make it clearly visible

---

## 🔹 Step 9: Add Chart Title
1. Select the chart
2. Go to **Format Pane (Paint Roller Icon)**
3. Expand **Title**
4. Turn **ON** Title
5. Set Title Text:Number of Employees by Gender

---

---

## 🔹 Step 10: Enable Data Labels
1. In **Format Pane**
2. Click **Data Labels**
3. Turn **ON**

✔ This shows count values on bars

---

## 🔹 Step 11: Change Colors (Optional but Recommended)
1. Go to **Format Pane**
2. Click **bars**
3. apply settings :
- Male → Blue
- Female → Pink

---

## 🔹 Step 12: Sort Chart (Optional)
1. Click on **3 dots (...)** on chart
2. Select **Sort by → Gender** or **Count of EmployeeID**
3. Choose Ascending/Descending

---

## 🚀 Alternative Method (Shortcut)

Instead of EmployeeID:
1. Drag **Gender → Y-axis**
2. Drag **Gender → X-axis**

✔ Power BI automatically counts values

---

##  Practice Task for Students
1. Create another bar chart:
- Axis → Department
- Values → Count of EmployeeID

2. Compare:
- Which department has highest employees?

---

# 📈 Power BI: Line Chart (Employee Trend) – Step by Step

## 🎯 Objective
Create a **Line Chart** to show how **employees are distributed over time**, so students can easily understand the **trend (increase/decrease)**.

---

## 🔹 Step 1: Create New Page
1. At the bottom, click **"+" (New Page)**
2. Rename page as:

   Line Chart

---

## 🔹 Step 2: Go to Report View
1. Click **Report View (chart icon on left)**

---

## 🔹 Step 3: Insert Line Chart
1. Go to **Visualizations Pane**
2. Click **Line Chart icon**

✔ Blank chart appears in center

---

## 🔹 Step 4: Add Date (X-axis)
1. Drag **JoinDate**
2. Drop into **X-axis**

---

## 🔹 Step 5: Fix Date Format (VERY IMPORTANT)
1. Click dropdown on **JoinDate (X-axis)**
2. Select:

   👉 **Date Hierarchy**

3. Keep only:
   - Month  
   OR  
   - Year

✔ Now X-axis shows time properly

---

## 🔹 Step 6: Add Values (Y-axis)
1. Drag **EmployeeID**
2. Drop into **Y-axis**

---

## 🔹 Step 7: Set Aggregation
1. Click dropdown beside **EmployeeID**
2. Select **Count**

✔ Now chart shows number of employees

---



## 🔹 Step 8: Resize Chart
1. Click chart
2. Drag corners to make it large

✔ Bigger chart = better understanding

---

## 🔹 Step 9: Add Title
1. Select chart
2. Go to **Format Pane (Paint Roller)**
3. Open **Title**
4. Turn **ON**
5. Enter:

   Employee Hiring Trend

---

## 🔹 Step 10: Turn ON Data Labels
1. Go to **Format Pane**
2. Click **Data Labels**
3. Turn **ON**

✔ Students can see exact values

---

## 🔹 Step 12: Enable Markers (Important for Learning)
1. Go to **Format Pane**
2. Click **Markers**
3. Turn **ON**

✔ Dots appear on line → easy to track values

---

## 🔹 Step 13: Change Line Color
1. Go to **Format Pane**
2. Click **Lines / Data Colors**
3. Choose a bright color (e.g., Blue)

---

## Task: Compare Male vs Female Trend

### Steps:
1. Drag **Gender**
2. Drop into **Legend**

✔ Now:
- Two lines appear
- Male vs Female comparison

---

# 🎯 Power BI: Gauge Chart (Target vs Actual)

## 🎯 Objective
Create a **Gauge Chart** to compare:
- **Actual Value** vs **Target Value**

👉 Example:
- Total Employees vs Target Employees

---

## 🔹 Step 1: Go to Report View
1. Click **Report View (Chart icon on left)**

---

## 🔹 Step 2: Insert Gauge Chart
1. Go to **Visualizations Pane**
2. Click **Gauge Chart icon**

✔ Blank gauge appears

---

## 🔹 Step 3: Add Value (Actual)
1. Drag **EmployeeID**
2. Drop into **Value**

---

## 🔹 Step 4: Set Aggregation
1. Click dropdown beside **EmployeeID**
2. Select **Count**

✔ Shows total employees

---

## 🔹 Step 5: Add Target Value

### Option 1: Manual Target
1. In **Fields pane**, click **New Measure**
2. Enter:Target Employees = 20



3. Press Enter

---

### Option 2: Use Existing Column
- Drag any numeric column into **Target Value**

---

## 🔹 Step 6: Assign Target
1. Drag **Target Employees**
2. Drop into **Target Value**

✔ Gauge now shows comparison

---

## 🔹 Step 7: Set Minimum & Maximum

1. Go to **Format Pane**
2. Find:
- Minimum → Set to 0
- Maximum → Set to Target (or higher)

✔ Example:
- Min = 0  
- Max = 20  

---

## 🔹 Step 8: Resize Chart
1. Click chart
2. Drag corners to adjust size

---

## 🔹 Step 9: Add Title
1. Select chart
2. Go to **Format Pane**
3. Open **Title**
4. Turn **ON**
5. Enter:

Employee Count vs Target

---

## 🔹 Step 10: Format Colors
1. Go to **Format Pane**
2. Find:
- Gauge fill color
- Target color

✔ Choose clear colors for visibility

---
